```python
# Variáveis e parâmetros
t = var('t')
m, g, kappa, T = var('m g kappa T')
Ix, Iy, Iz, Ixz = var('I_x I_y I_z I_xz')
X, Y, Z, L, M, N = var('X Y Z L M N')

# Estados dependentes do tempo
U, V, W, p, q, r, theta, phi = [function(n)(t) for n in
    ['U', 'V', 'W', 'p', 'q', 'r', 'theta', 'phi']]

# Equações de força
eq1 = m*(diff(U,t) + q*W - r*V) == X - m*g*sin(theta) + kappa*T*cos(theta)
eq2 = m*(diff(V,t) + r*U - p*W) == Y + m*g*cos(theta)*sin(phi)
eq3 = m*(diff(W,t) + p*V - q*U) == Z + m*g*cos(theta)*cos(phi) - kappa*T*sin(theta)

# Equações de momento
eq4 = Ix*diff(p,t) + Ixz*diff(r,t) + (Iz - Iy)*q*r + Ixz*q*p == L
eq5 = Iy*diff(q,t) + (Ix - Iz)*p*r + Ixz*(r^2 - p^2) == M
eq6 = Iz*diff(r,t) + Ixz*diff(p,t) + (Iy - Ix)*q*p - Ixz*q*r == N

eqs = [eq1, eq2, eq3, eq4, eq5, eq6]
for e in eqs:
    show(e)
```

```python
dp, dr = var('dp dr')

# Substitui as derivadas por símbolos
e4 = eq4.subs({diff(p,t): dp, diff(r,t): dr})
e6 = eq6.subs({diff(p,t): dp, diff(r,t): dr})

sol = solve([e4, e6], dp, dr, solution_dict=True)[0]

dp_expr = sol[dp].simplify_full()
dr_expr = sol[dr].simplify_full()

show(LatexExpr(r"\dot p = "), dp_expr)
show(LatexExpr(r"\dot r = "), dr_expr)
```
